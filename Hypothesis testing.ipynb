One-Sample t-Test
A school claims that the average score of students in a test is 75. A random
sample of 10 students' scores is collected, and you want to test if the sample
mean significantly differs from the claimed population mean (α=0.05)

```python
import numpy as np
import scipy.stats as stats
```

```python
scores = [78, 72, 74, 76, 73, 79, 77, 71, 70, 74]
population_mean = 75
n = len(scores)
```

```python
sample_mean = np.mean(scores)
sample_std = np.std(scores,
ddof=1)
```

```python
t_stat = (sample_mean - population_mean) / (sample_std / np.sqrt(n))
```

```python
df = n - 1
p_value = 2 *stats.t.sf(abs(t_stat), df)
print(f"Sample Mean:{sample_mean:.2f}")
print(f"tstatistic: {t_stat:.2f}")
print(f"p-value: {p_value:.4f}")
if p_value < 0.05:
  print("Reject the null hypothesis: The sample mean is significantly different.")
else:
  print("Fail to reject the null hypothesis: No significant difference.")
```

Z-test
A factory claims the average weight of a product is 500g. A sample of 40 products
has a mean weight of 495g, and the known population standard deviation is 10g.
Test the claim (α=0.05).


```python
import numpy as np
import scipy.stats as stats
```

```python
sample_mean = 495
population_mean = 500
population_std = 10
n = 40
z_stat = (sample_mean - population_mean) / (population_std / np.sqrt(n))
p_value = 2 *stats.norm.sf(abs(z_stat))

print(f"z-statistic:{z_stat:.2f}")
print(f"pvalue: {p_value:.4f}")
if p_value < 0.05:
  print("Reject the null hypothesis: The sample mean is significantly different.")
else:
  print("Fail to reject the null hypothesis: No significant difference")


```
