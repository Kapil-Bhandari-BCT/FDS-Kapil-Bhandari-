```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
sns.set()
```

```python
%config InlineBackend.figure_format = 'svg'
DATA_URL = "https://raw.githubusercontent.com/Yorko/mlcourse.ai/main/data/"
data_url = DATA_URL + "telecom_churn.csv"
df = pd.read_csv(data_url)
```

```python
df.head()
```

```python
selected_features = ["Total day minutes", "Total intl calls"]
```

```python
df[selected_features].hist(figsize=(10, 4))
```

```python
df[selected_features].plot(
kind="density",
subplots=True,
layout=(1, 2),
sharex=False,
figsize=(10, 4)
)
```

```python
sns.histplot(df["Total intl calls"], kde=True, stat="density");
```

```python
sns.boxplot(x="Total intl calls", data=df);
```

```python
numerical = list(
set(df.columns)
- set(
[
"State",
"International plan",
"Voice mail plan",
"Area code",
"Churn",
"Customer service calls",
]
)
)
```

```python
corr_matrix = df[numerical].corr()
sns.heatmap(corr_matrix);
```

```python
plt.scatter(df["Total day minutes"], df["Total night minutes"]);
```

```python
sns.jointplot(x="Total day minutes", y="Total night minutes", data=df, kind="scatter");
```
